```python
import numpy as np
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
X = np.array([[-1, -1], [-2, -1], [1, 1], [2, 1]])
y = np.array([1, 1, 2, 2])
from sklearn.svm import SVC
clf = make_pipeline(StandardScaler(), SVC(gamma='auto'))
clf.fit(X, y)
```

```python
print(clf.predict([[-0.8, -1]]))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Scatter plot of X vs y labels (second feature vs first feature)
# Assumes X and y are already defined in previous cells.
plt.figure(figsize=(5,4))
classes = np.unique(y)
for idx, cls in enumerate(classes):
    plt.scatter(X[y==cls, 0], X[y==cls, 1], s=70, label=f"class {cls}")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Scatter plot of X by class label y")
plt.legend()
plt.grid(True)
plt.show()
```
